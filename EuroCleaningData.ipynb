```python
# Team,Ball possession,Big chances,Total shots,Goalkeeper saves,Corner kicks,Fouls,Passes,Tackles,Free kicks,Yellow cards,Red cards,Shots on target,Hit woodwork,Shots off target,Blocked shots,Shots inside box,Shots outside box,Big chances scored,Big chances missed,Through balls,Touches in penalty area,Fouled in final third,Offsides,Accurate passes,Throw-ins,Final third entries,Long balls,Crosses,Duels,Dispossessed,Ground duels,Aerial duels,Dribbles,Tackles won,Total tackles,Interceptions,Recoveries,Clearances,Goals prevented,Goalkeeper saves,Goal kicks,Goals


# combining the game stats into one row

def combine_game_stats(file_path, target_team, output_file):
    stat_columns =  stat_columns = ['Ball possession', 'Big chances', 'Total shots', 'Goalkeeper saves', 'Corner kicks', 'Fouls',
                    'Passes', 'Tackles', 'Free kicks', 'Yellow cards', 'Red cards', 'Shots on target', 
                    'Hit woodwork', 'Shots off target', 'Blocked shots', 'Shots inside box', 'Shots outside box', 
                    'Big chances scored', 'Big chances missed', 'Through balls', 'Touches in penalty area', 
                    'Fouled in final third', 'Offsides', 'Accurate passes', 'Throw-ins', 'Final third entries', 
                    'Long balls', 'Crosses', 'Duels', 'Dispossessed', 'Ground duels', 'Aerial duels', 'Dribbles', 
                    'Tackles won', 'Total tackles', 'Interceptions', 'Recoveries', 'Clearances', 'Goals prevented', 
                    'Goalkeeper saves', 'Goal kicks', 'Goals']
    df = pd.read_csv(file_path, encoding ='latin1')
    grouped= df.groupby('Game Link')

    combined_data = []
    for game_link, group in grouped:
        if len(group) == 2:
            team1, team2 = group.iloc[0], group.iloc[1]

            if team1['Team'] == target_team:
                target_row = team1
                opponent_row = team2
            else:
                target_row = team2
                opponent_row = team1
            combined_row = {'Game Link': game_link,
                            'Target Team': target_row['Team'],
                            'Opponent': opponent_row['Team'],
                            'Date': target_row['Date'],}
            for col in stat_columns:
                combined_row[f'{col} For'] = target_row[col]
                combined_row[f'{col} Against'] = opponent_row[col]
            combined_data.append(combined_row)
        combined_df = pd.DataFrame(combined_data)
        combined_df.to_csv(output_file,index = False)
    return combined_df

file_path = '/Users/grealish/Desktop/Euros Prediction Model/EuroPrediction/Game Statistics/Albania_ALL.csv' 
output_file = '/Users/grealish/Desktop/Euros Prediction Model/EuroPrediction/Game Stats Cleaned/Albania_Combined_Performance.csv'  
target_team = 'Albania'

combined_df = combine_game_stats(file_path, target_team, output_file)

combined_df
```

```python
# putting all of the documents into Game Stats Cleaned:

input_dir = '/Users/grealish/Desktop/Euros Prediction Model/EuroPrediction/Game Statistics'
output_dir = '/Users/grealish/Desktop/Euros Prediction Model/EuroPrediction/Game Stats Cleaned'

#team names without turkey
team_names = ['Austria','Belgium','Croatia','Czechia','Denmark','England','France','Georgia','Germany','Hungary','Italy','Netherlands',
              'Poland','Portugal','Romania','Scotland','Serbia','Slovakia','Slovenia','Spain','Switzerland','Ukraine'] 

for team in team_names:
    file_path = os.path.join(input_dir, f'{team}_ALL.csv')

    output_file = os.path.join(output_dir,f'{team}_Combined_Performance.csv')
    try:
        combine_game_stats(file_path,team,output_file)
        print(f'Processed {team} successfully.')
    except FileNotFoundError:
        print(f'File for {team} not found. Skipping...')
    except Exception as e:
        print(f'An error occurred while prcessing {team}: {e}')
```
